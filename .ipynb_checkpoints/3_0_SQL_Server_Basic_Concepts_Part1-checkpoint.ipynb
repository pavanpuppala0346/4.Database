### 1. Intro to DBMS
<b>Language</b>

Advantage: 
- Process the Data
- Early programming languages like C or assembly were able to process raw data stored in memory or files.
- You could write code to read, manipulate, and calculate results from data.
- Example: Reading a text file of student marks and calculating averages.
- Advantage: You had full control over how the data is handled.

Disadvantage: 
- Cannot Store Data
- Programming languages themselves do not provide a built-in mechanism to store data permanently.
- Whatever data is processed resides in memory (RAM), and once the program ends or the computer shuts down, the data is lost.
- Permanent storage had to be handled through flat files or external systems.

<b>Issues with Flat File</b>

Flat files are simple text/binary files used to store data (like .txt or .csv). While easy to use, they have many problems when data volume and complexity increase.

1. Size Issues
    - Flat files grow quickly as data increases.
    - Example: A sales company storing all invoices in one CSV file — once the file reaches gigabytes of size, it becomes slow and unmanageable.
    - Searching, updating, or deleting records takes too long.

2. Security
    - Flat files usually lack access control. Anyone with access to the file system can open and modify the file.
    - No encryption or role-based access like databases provide.
    - Sensitive data (e.g., salaries, credit card numbers) is at risk.

3. Retrieve the Data → Slow
    - To find one record, the system must scan the entire file sequentially (linear search).
    - Example: Finding one customer’s record in a file of 10 million rows could take a very long time.

4. C Language → Complex
    - Writing file handling code in C is complex and error-prone.
    - Example: You must manually handle memory allocation, buffer overflow issues, and error handling.
    - No built-in easy-to-use functions like in modern databases.

5. No Indexes → Processing Slow
    - Databases use indexes (like an index in a book) to quickly locate data.
    - Flat files don’t support indexing. If you want to search for a record, you must read the file line by line.
    - Example: Searching for Employee ID = 100000 in a 5 GB file → system reads every record until it finds it.

6. Concurrency → Multiple Customers Not Allowed
    - In real-world applications (like banking), multiple users need to access data at the same time.
    - Flat files can only be accessed by one process at a time; if multiple users try, data corruption may occur.
    - Example: If two customers withdraw money at the same time from a flat file, one transaction may overwrite the other → data inconsistency.

### 34. Raking Functions and CTE

- These are used to provide ranking for rows based on one or more conditions.
- Ranking Functions
	- Row_Number()   : Returns the sequential number starting from 1
	- Rank()         : Returns the rank for each row, with Gap
	- Dense_Rank()   : Returns the rank for each row without gaps
	- NTILE()        : splits the data into ‘n’ no. of groups Assigns an unique number to each group starting from 1
	- Partition By   : Groups the data
- Introduced in SQL Server 2005.
  
Syntax: 

```sql  
    SELECT Columns,
           Rank FunctionName()
           OVER (ORDER BY COLUMN DESC/ASC)
           As Alias
	FROM TABLE
```
<b>Using row_number()</b>
```sql
SELECT ACID, NAME, CBAL, BRID, ROW_NUMBER() OVER (ORDER BY ACID DESC) AS RNO FROM AMASTER
```

<b>get the 5th Row</b>
```sql
SELECT * FROM (
    SELECT ACID, NAME, CBAL, BRID, ROW_NUMBER() OVER (ORDER BY ACID DESC) AS RNO FROM AMASTER
    ) as k
WHERE RNO=5
```

<b>get Rows between 5 and 10th </b>
```sql
SELECT * FROM (
    SELECT ACID, NAME, CBAL, BRID, ROW_NUMBER() OVER (ORDER BY ACID DESC) AS RNO FROM AMASTER
    ) as k
WHERE RNO BETWEEN 5 and 10
```

<b>Using partition by()</b>
```sql
SELECT ACID, NAME, CBAL, BRID, ROW_NUMBER() OVER (Partition by BRID ORDER BY ACID ASC) AS RNO FROM AMASTER
    
SELECT ACID, NAME, CBAL, BRID, ROW_NUMBER() OVER (Partition by BRID ORDER BY ACID DESC) AS RNO FROM AMASTER
```

<b>first customer from each branch</b>
```sql
SELECT ACID, NAME, CBAL, BRID FROM (
    SELECT ACID, NAME, CBAL, BRID, ROW_NUMBER() OVER (Partition by BRID ORDER BY ACID ASC) AS RNO FROM AMASTER
    ) as K
WHERE RNO=1
```

<b>IQ: Every 5th customer from each branch</b>
```sql
SELECT ACID, NAME, CBAL, BRID FROM (
    SELECT ACID, NAME, CBAL, BRID, ROW_NUMBER() OVER (ORDER BY ACID ASC) AS RNO FROM AMASTER
    ) as K
WHERE RNO % 5 = 0
```

<b>Alternate customer from each branch</b>
```sql
SELECT ACID, NAME, CBAL, BRID FROM (
    SELECT ACID, NAME, CBAL, BRID, ROW_NUMBER() OVER (ORDER BY ACID ASC) AS RNO FROM AMASTER
    ) as K
```


#### Common Table Expression(CTE)
-	It is similar to Derived table(a query in the FROM Clause)
-	CTE is introduced in SQL Server 2005

Syntax: 
```sql

	     select statement
	)
```


### 35.derived tables vs CTE vs Temp Tables

<b>Find out the branch which has highest number of customers using Dervied table</b>

```sql
SELECT BRID 
FROM (
        SELECT BRID, count(*) as CNT
        FROM AMASTER
        GROUP BY BRID
    )k
WHERE CNT = (
                SELECT max(CNT) 
                FROM (
                        SELECT BRID, count(*) as CNT
                        FROM AMASTER
                        GROUP BY BRID
                     )k
            )
```
<b>Find out the branch which has highest number of customers using CTE</b>

```sql
with k 
as 
(
    SELECT BRID, count(*) as CNT
    FROM AMASTER
    GROUP BY BRID
)

SELECT *
FROM k
WHERE CNT = (
                SELECT max(CNT) 
                FROM k
            )
```

<b>Find out the branch which has highest number of customers using TEMP TABLE</b>

```sql
SELECT BRID, count(*) as CNT into #K
FROM AMASTER
GROUP BY BRID

SELECT BRID 
FROM #K
WHERE CNT=( SELECT max(CNT)  FROM #K )

SELECT BRID 
FROM #K
WHERE CNT=( SELECT max(CNT)  FROM #K WHERE CNT<( ELECT max(CNT)  FROM #K ))
```

#### TEMP Tables are of 2 types
- Local temp table   #k
- Global temp table  ##k


|  Derived Tables     |   CTE               | Local Temp Tables   |Global Temp Tables         |
|---------------------|---------------------|---------------------|---------------------------|
| Created Once        | Created Once        | Created Once        | Created Once              |
| Use Once            | Use Many Times      | Use Many Times      | Use Many Times            |
| Scope: Single Query | Scope: Single Query | Scope: Window Level | Scope: Across All Windows |

```sql

```
<b>How to remove duplicate rows from table using CTE</b>
```sql 
with k
as(
    SELECT name, row_number() over (PARTITION BY name Order by name asc) as Rno FROM AMASTER 
)

delete from k where Rno > 1
```

<b>get the Running total within the branch of AMASTER table using WINDOW FUNCTION</b>
```sql

```

<b>get the Running total within the branch of AMASTER table using JOINS</b>
```sql 
SELECT M1.acid, M1.name, M1.cbal, M1.BRID, sum(M1.cbal) as Running_Total 
FROM AMASTER as M1
INNER JOIN AMASTER as M2 on M1.acid<=m2.acid
GROUP BY M1.acid, M1.name, M1.cbal, M1.BRID
ORDER BY M1.acid
```

i have a table Sales and column SalesAmt 1,2,3,-1,-2,-3 i want the output of sum of positive number and sum of negative numbers in 2 different columns

| SalesAmt |
| -------- |
| 1        |
| 2        |
| 3        |
| -1       |
| -2       |
| -3       |

Output:
| SumPositive | SumNegative |
| ----------- | ----------- |
| 6           | -6          |

```sql
SELECT
    SUM(CASE WHEN SalesAmt > 0 THEN SalesAmt ELSE 0 END) AS SumPositive,
    SUM(CASE WHEN SalesAmt < 0 THEN SalesAmt ELSE 0 END) AS SumNegative
FROM Sales;
```

### 36. merge statement to perform incremental loading
-	To perform incremental loading
-	It is one of the DML statement placed newly in SQL server 2008
-	Merge statement is a combination statement that can perform INSERT, UPDATE, DELETE statements based on whether rows that match the selection criteria exist in the target table or not

Syntax:  
```sql
     MERGE target_table as target
     USING source_table as source
     ON condition 
     WHEN Matched THEN 
     UPDATE
     SET column = source.column,………….
     WHEN target not matched  THEN 
     INSERT VALUES(source.column1, source.column2……)
```

#### Case Studies
- Everyday we delete the data and do the full copy in target table this is applicable only we have less number of rows but is not good practice for large datasets Because may be some applications dependent on target tables
- To Overcome above problem we use Incremental Loading

```sql
    Merge DEST_STUDENT as t
    using SRC_STUDENT as s
    on t.eid = s.eid

    when Not matched by Target
    then
    insert into DEST_STUDENT(id, name, salary, DT_START) values(s.id, s.name, s.salary, getdate())

    when matched and t.name <> s.name or t.salary <> s.salary
    then
    update DEST_STUDENT
    set t.name = s.name, t.salary = s.salary, t.DT_START = getdate()
        
    when Not matched by Source
    then delete
```

```sql
    alter table DEST_STUDENT
    add DT_START datetime;  
```
