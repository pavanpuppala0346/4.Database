#### 2. to retireve the data

```python
res = col.find({})
for r in res:
    print(r)
```

```python
res = col.find({}).limit(3)
for r in res:
    print(r)
```

#### to retrieve only wanted columns

```python
res = col.find({},{'name','phone'})
for r in res:
    print(r)
```

#### retrieve data based on some condition

```python
res = col.find({'name':'pawan'})
for r in res:
    print(r)
```

```python
res = col.find({'age':{'$gt':'24'}})
for r in res:
    print(r)
```

```python
query = {'age':{'$gt':'24'}}
res = col.find(query).sort('name')
for r in res:
    print(r)
```

```python
query = {'age':{'$gt':'24'}}
res = col.find(query).sort('name',-1)
for r in res:
    print(r)
```
